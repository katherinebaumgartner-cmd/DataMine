## Question 4

```r
my_flights <- read.csv('/anvil/projects/tdm/data/flights/1997.csv')
```

```r
breaks <- c(0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400)
```

```r
labels <- c("Midnight–3AM", "3AM–6AM", "6AM–9AM", "9AM–12PM", "12PM–3PM", "3PM–6PM", "6PM–9PM", "9PM–12AM")
```

```r
my_flights$depart_times <- cut(my_flights$DepTime, breaks = breaks, labels = labels, include.lowest = TRUE)
```

```r
barplot(table(my_flights$depart_times), main = "Flight Departures by Time of Day (1997)", xlab = "Departure Time Range", ylab = "Number of Flights", las = 2)
```

```r
my_flights$arrival_times <- cut(my_flights$ArrTime, breaks = breaks, labels = labels, include.lowest = TRUE)

barplot(table(my_flights$arrival_times), main = "Flight Arrivals by Time of Day (1997)", xlab = "Arrival Time Range", ylab = "Number of Flights", las = 2)
```

DELIVERABLES
4.1) 6am-9am
4.2) 12am-3am
4.3) My conclusion is that flight times are moslty clustered around the early morning hours and the late night flights are rarer.
