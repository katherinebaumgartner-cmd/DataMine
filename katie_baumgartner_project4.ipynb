```r
my_vec1 <- myDF$Age[!is.na(myDF$Age) & myDF$Age <= 10]
my_vec2 <- myDF$Age[!is.na(myDF$Age) & myDF$Age > 10 & myDF$Age <= 60]
```

```r
people <- c(my_vec1, my_vec2)
```

```r
children_vec <- people[people <= 20]
young_vec <- people[people > 20 & people <= 40]
adult_vec <- people[people > 40 & people <= 60]
```

```r
AgeGroup <- ifelse(myDF$Age %in% children_vec, "Child", ifelse(myDF$Age %in% young_vec, "Young", ifelse(myDF$Age %in% adult_vec, "Adult", NA)))
```

```r
newVar <- table(AgeGroup, myDF$Combined)
```

```r
# Make room for long labels
par(mar = c(10, 5, 4, 2))

barplot(
  newVar,
  beside = TRUE,
  col = rainbow(nrow(newVar)), 
  las = 2,                                   
  main = "Survival Status by Age Group and Sex",
  xlab = "Combined Categories",
  ylab = "Count"
)

legend(
  "topright",
  legend = rownames(newVar),
  fill = rainbow(nrow(newVar)),
  title = "Age Group"
)
```
